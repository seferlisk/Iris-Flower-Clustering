```python
from sklearn.datasets import load_iris
import pandas as pd

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print(df.head(5))
```

```python
# We map the integers (0,1,2) to actual names for better readability in the plot
df['Species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

print(df.head(5))
```
